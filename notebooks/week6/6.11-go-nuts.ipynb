```python
from pathlib import Path
import runpy
root = Path.cwd()
while not (root / "data/week5/marvel_pages.zip").exists():
    if root == root.parent:
        raise RuntimeError("Open the notebook from inside the course repository")
    root = root.parent
builder = runpy.run_path(str(root / "notebooks/week6/build_week6.py"))
result = builder["build"]()
result["summary"]
```

## Inspect the evidence
Scores are calculated in the full TF-IDF space. Each shared-term contribution is the product of two normalized TF-IDF weights. A disconnected pair has `None` distance.

```python
for pair in result["examples"][:5]:
    print(result["nodes"][pair["a"]]["name"], "↔", result["nodes"][pair["b"]]["name"])
    print("Cosine:", pair["cosine"], "hops:", pair["distance"], "name-filtered cosine:", pair["clean_cosine"])
    print("Shared terms:", [t["word"] for t in pair["shared"]])
    print(*pair["excerpts"], sep="\n")
    print()
```
