# Week 3 · Six cliques are one X-Men core in disguise

This notebook is the short, inspectable entry point for the analysis behind the Week 3 post. The pipeline preserves the frozen roster, collapses direction for the clique question, enumerates maximum cliques, and compares them with a degree-preserving null.

```python
from pathlib import Path
import importlib.util

script = Path('build_week3.py')
if not script.exists():
    script = Path('notebooks/week3/build_week3.py')
spec = importlib.util.spec_from_file_location('week3_builder', script)
week3 = importlib.util.module_from_spec(spec)
spec.loader.exec_module(week3)
```

```python
artifact = week3.build_artifact()
artifact['network'], {k: v for k, v in artifact['null'].items() if k != 'max_clique_numbers'}
```

```python
assert artifact['network']['clique_number'] == 8
assert artifact['network']['maximum_cliques'] == 6
assert len(artifact['core']) == 6
assert sorted([len(artifact['seat_a']), len(artifact['seat_b'])]) == [2, 3]
assert artifact['null']['max_clique_max'] < artifact['network']['clique_number']
```

## Interpretation

All six maximum cliques share the same six-character core. Their remaining members are one choice from a two-character set and one from a three-character set, which produces exactly six lineups. None of the 100 degree-preserving shuffles reaches the observed clique number of eight.
